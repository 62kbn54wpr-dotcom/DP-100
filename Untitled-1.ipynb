# Preparation of the System

## Installation of Azure CLI
Installation of Azure CLI throough running the following command from the system Terminal
> brew update && install azure-cli

## Installation of .env to keep secrets out of the code
Run the following command from the .venv Terminal
> pip3 install python-dotenv

Create a .txt file with the secrets in to in the form
>Key (secret) = Value

For Azure, the following secrets may be of interest:
- AZURE_RESOURCE_GROUP_NAME # Preferred resource group name
- LOCATION # Preferred Azure region
- AZURE_SUBSCRIPTION_ID # subscription ID obtained through 'az account show --query id --output tsv'

## Create .venv
Use the latest Python version installed through brew

Omit both .env and .venv from replication through GitHub
