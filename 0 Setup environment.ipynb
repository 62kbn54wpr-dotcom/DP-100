# Preparation of the System

## 1 Prerequisites
### 1.1 Installation of Azure CLI
Installation of Azure CLI throough running the following command from the system Terminal
> brew update && install azure-cli

### 1.2 Create directory in GitHub

Create a repository in GitHub and clone the repository locally.

Create a .gitignore file in the local repository.

Add .DS_Store in the .gitignore file to prevent the replication of .DS_Store file generated by MacOS.

The local repository will be used to create the .venv environment

### 1.3 Create .venv
Use the latest installed (brew) Python version.

Omit .venv from GitHub replication by adding .venv in the .gitignore file

### 1.4 Installation of .env to keep secrets out of the code
Run the following command from the .venv Terminal

> pip3 install python-dotenv

## 3 Setup environment variables (secrets)
In this step, you set environment variables for use in the code.  We use dotenv to retreive the variables later on.

Create a .env file with the secrets in to in the form

>Key (secret) = Value

For Azure, the following secrets may be of interest:

- AZURE_RESOURCE_GROUP_NAME # Preferred resource group name

- LOCATION # Preferred Azure region

- AZURE_SUBSCRIPTION_ID # subscription ID obtained through 'az account show --query id --output tsv'

Omit the .env from GitHub replication by adding .env in the .gitignore file
